```python
import numpy as np
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.metrics import classification_report
```

```python
df = pd.read_excel("grokData.xlsx")
```

```python
ref_col = 'Hop_7_Impact_Severity_Level (Ref)'
pred_col = 'Hop_7_Impact_Severity_Level'
abs_col = 'Ablation Severity'
```

```python
df = df[df["Hop_1_Triage (Ref)"] != "بی‌ضرر"]
df = df[df["Hop_1_Triage"] != "بی‌ضرر"]
df[pred_col] = df[pred_col].fillna('')
print(len(df))
```

```python
print(len(df[ref_col].unique()))
```

```python
print(len(df[pred_col].unique()))
```

```python
print(len(df[abs_col].unique()))
```

```python
precision = precision_score(df[ref_col], df[pred_col], average='macro')
recall    = recall_score(df[ref_col], df[pred_col], average='macro')
f1        = f1_score(df[ref_col], df[pred_col], average='macro')
```
