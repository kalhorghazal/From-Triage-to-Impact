```python
import numpy as np
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.metrics import classification_report
from sklearn.metrics import cohen_kappa_score
```

```python
df = pd.read_excel("o3Data.xlsx")
```

```python
ref_col = 'Hop_7_Impact_Severity_Level (Ref)'
pred_col = 'Hop_7_Impact_Severity_Level'
abs_col = 'Ablation Severity'
```

```python
df = df[df["Hop_1_Triage (Ref)"] != "بی‌ضرر"]
df[pred_col] = df[pred_col].fillna('')
```

```python
print(len(df[ref_col].unique()))
```

```python
print(len(df[pred_col].unique()))
```

```python
print(len(df[abs_col].unique()))
```

```python
precision = precision_score(df[ref_col], df[pred_col], average='macro', zero_division=0)
recall    = recall_score(df[ref_col], df[pred_col], average='macro', zero_division=0)
f1        = f1_score(df[ref_col], df[pred_col], average='macro', zero_division=0)
```

```python
severity_order = {
    'کم': 0,
    'متوسط': 1,
    'زیاد': 2,
    'شدید': 3
}

df_eval = df[df[pred_col].isin(severity_order.keys())]

y_ref = df_eval[ref_col].map(severity_order)
y_pred = df_eval[pred_col].map(severity_order)

weighted_kappa = cohen_kappa_score(
    y_ref,
    y_pred,
    weights='quadratic'
)
```

## Ablation Study

```python
precision2 = precision_score(df[ref_col], df[abs_col], average='macro')
recall2    = recall_score(df[ref_col], df[abs_col], average='macro')
f12        = f1_score(df[ref_col], df[abs_col], average='macro')
```

```python
y_abs = df_eval[abs_col].map(severity_order)

weighted_kappa2 = cohen_kappa_score(
    y_ref,
    y_abs,
    weights='quadratic'
)
```
