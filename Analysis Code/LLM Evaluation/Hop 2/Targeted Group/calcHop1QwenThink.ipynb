```python
import numpy as np
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.metrics import classification_report
```

```python
df = pd.read_excel("qwenThinkData.xlsx")
```

```python
ref_col = 'Hop_1_Triage (Ref)'
pred_col = 'Hop_1_Triage'
```

```python
print(len(df[ref_col].unique()))
```

```python
print(len(df[pred_col].unique()))
```

```python
precision = precision_score(df[ref_col], df[pred_col], average='macro')
recall    = recall_score(df[ref_col], df[pred_col], average='macro')
f1        = f1_score(df[ref_col], df[pred_col], average='macro')
```

```python
precision
```

```python
recall
```

```python
f1
```

```python
print(classification_report(df[ref_col], df[pred_col]))
```
