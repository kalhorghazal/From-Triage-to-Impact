```python
import numpy as np
import pandas as pd
import ast
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.metrics import f1_score
```

```python
df = pd.read_excel("humanData.xlsx")
```

```python
annotator1_col = 'Hop_4_Moral_Foundation (A1)'
annotator2_col = 'Hop_4_Moral_Foundation (A2)'
```

```python
df = df[df["Hop_1_Triage (A1)"] != "بی‌ضرر"]
df = df[df["Hop_1_Triage (A2)"] != "بی‌ضرر"]
print(len(df))
```

```python
annotator1Morals = []
for annotator1Moral in df[annotator1_col]:
  annotator1Morals.extend(ast.literal_eval(annotator1Moral))
set(annotator1Morals)
```

```python
annotator2Morals = []
for annotator2Moral in df[annotator2_col]:
  annotator2Morals.extend(ast.literal_eval(annotator2Moral))
set(annotator2Morals)
```

```python
df[annotator1_col] = df[annotator1_col].apply(ast.literal_eval)
df[annotator2_col] = df[annotator2_col].apply(ast.literal_eval)
```

```python
mlb = MultiLabelBinarizer()

mlb.fit(list(df[annotator1_col]) + list(df[annotator2_col]))

y_true = mlb.transform(df[annotator1_col])
y_pred = mlb.transform(df[annotator2_col])
```

```python
f1_micro = f1_score(y_true, y_pred, average='micro')
f1_macro = f1_score(y_true, y_pred, average='macro')

print(f"Micro F1: {f1_micro:.4f}")
print(f"Macro F1: {f1_macro:.4f}")
```
