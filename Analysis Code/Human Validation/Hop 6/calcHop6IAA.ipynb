```python
import numpy as np
import pandas as pd
from bert_score import score as bert_score
from sacrebleu.metrics import CHRF
from rouge_score import rouge_scorer
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_excel("humanData.xlsx")
```

```python
annotator1_col = 'Hop_6_Narrative (A1)'
annotator2_col = 'Hop_6_Narrative (A2)'
```

```python
df = df[df["Hop_1_Triage (A1)"] != "بی‌ضرر"]
df = df[df["Hop_1_Triage (A2)"] != "بی‌ضرر"]
print(len(df))
```

## BERTScore-F1

```python
P, R, F1 = bert_score(
    df[annotator1_col].tolist(),
    df[annotator2_col].tolist(),
    model_type="HooshvareLab/bert-base-parsbert-uncased",
    num_layers=12,
    lang="fa",
    verbose=True
)

bert_score_f1 = F1.mean().item()

print(f"BERTScore-F1: {bert_score_f1:.4f}")
```

```python
print(f"BERTScore-F1: {bert_score_f1:.4f}")
```

## chrF++

```python
chrf = CHRF(word_order=2)

chrf_scores = []

for a1, a2 in zip(df[annotator1_col], df[annotator2_col]):
    score = chrf.sentence_score(a2, [a1])
    chrf_scores.append(score.score)

chrf_score = np.mean(chrf_scores)

print(f"chrF++: {chrf_score:.4f}")
```

## Cosine similarity

```python
model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
)

ref_embeddings = model.encode(
    df[annotator1_col].tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

pred_embeddings = model.encode(
    df[annotator2_col].tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

cosine_scores = []

for ref_embedding, pred_embedding in zip(ref_embeddings, pred_embeddings):
    score = cosine_similarity(
        [ref_embedding],
        [pred_embedding]
    )[0][0]

    cosine_scores.append(score)

cosine_similarity_score = np.mean(cosine_scores)

print(f"Cosine Similarity: {cosine_similarity_score:.4f}")
```
