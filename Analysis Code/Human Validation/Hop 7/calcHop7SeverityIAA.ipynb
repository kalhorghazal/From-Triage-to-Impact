```python
import numpy as np
import pandas as pd
from sklearn.metrics import cohen_kappa_score
```

```python
df = pd.read_excel("humanData.xlsx")
```

```python
annotator1_col = 'Hop_7_Impact_Severity_Level (A1)'
annotator2_col = 'Hop_7_Impact_Severity_Level (A2)'
```

```python
df = df[df["Hop_1_Triage (A1)"] != "بی‌ضرر"]
df = df[df["Hop_1_Triage (A2)"] != "بی‌ضرر"]
print(len(df))
```

```python
for x in df[annotator1_col].unique():
  print(x)
```

```python
for x in df[annotator2_col].unique():
  print(x)
```

```python
a1 = df[annotator1_col]
a2 = df[annotator2_col]
```

```python
severity_order = {
    'کم': 0,       # Low
    'متوسط': 1,    # Medium
    'زیاد': 2,     # High
    'شدید': 3      # Severe
}
```

```python
a1_num = a1.map(severity_order)
a2_num = a2.map(severity_order)

kappa_weighted = cohen_kappa_score(
    a1_num,
    a2_num,
    weights='quadratic'
)

print(f"Weighted Cohen's kappa: {kappa_weighted:.4f}")
```

```python
agreement = (a1_num == a2_num).mean()
print(f"Observed agreement: {agreement:.4f}")
```
