```python
import numpy as np
import pandas as pd
from sklearn.metrics import cohen_kappa_score
```

```python
df = pd.read_excel("humanData.xlsx")
```

```python
annotator1_col = 'Hop_1_Triage (A1)'
annotator2_col = 'Hop_1_Triage (A2)'
```

```python
for x in df[annotator1_col].unique():
  print(x)
```

```python
for x in df[annotator2_col].unique():
  print(x)
```

```python
a1 = df[annotator1_col]
a2 = df[annotator2_col]
```

```python
kappa = cohen_kappa_score(a1, a2)
print(f"Cohen's kappa: {kappa:.3f}")
```

```python
agreement = (a1 == a2).mean()
print(f"Observed agreement: {agreement:.3f}")
```
